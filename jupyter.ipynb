BUILD MASTER MERGED DATAFRAME

```python


# start with orders as the backbone table (one row per order)
master = orders.copy()

# merge customers - adds customer_unique_id (needed for segmentation later, not customer_id)
master = master.merge(customers, on='customer_id', how='left')

# merge order_items - this is ITEM level, so row count will increase here (some orders have multiple items)
master = master.merge(order_items, on='order_id', how='left')

# merge products - adds category name (Portuguese) + product metadata
master = master.merge(products, on='product_id', how='left')

# translate category name from Portuguese to English using the lookup table
master = master.merge(category_translation, on='product_category_name', how='left')

# merge sellers - adds seller location info
master = master.merge(sellers, on='seller_id', how='left')

# merge payments_agg (already aggregated to order-level in an earlier step) - safe, won't duplicate rows
master = master.merge(payments_agg, on='order_id', how='left')

# merge reviews_clean (deduped by order_id in an earlier step) - safe, won't duplicate rows
reviews_clean = reviews.drop_duplicates(subset='order_id', keep='first')
master = master.merge(reviews_clean, on='order_id', how='left')

# check final shape - should be close to order_items row count (~112,650) since that's the finest grain (item-level)
print("Master dataframe shape:", master.shape)
master.head()
```

```python
# sanity check: confirm row count matches expectation (item-level, so should be close to order_items shape)
print("order_items shape:", order_items.shape)
print("master shape:", master.shape)

# check for any unexpected new missing values introduced by the merges
print(master.isnull().sum()[master.isnull().sum() > 0])
```

```python
# check how many orders have zero matching order_items rows
# these become the "extra" rows in master with NaN item-level data
orphan_orders = master[master['order_item_id'].isnull()]
print("Orders with no items:", orphan_orders.shape[0])
print(orphan_orders['order_status'].value_counts())
```

```python
# fill missing category with 'unknown' - covers both orphan orders AND unmatched product_ids
# doing this at master level (not just on the products table) since orphan rows introduce fresh NaNs too
master['product_category_name'] = master['product_category_name'].fillna('unknown')
master['product_category_name_english'] = master['product_category_name_english'].fillna('unknown')

# leave price, freight_value, product dimensions, review dates as NaN - these are genuine 
# "no item/product/review exists for this order" cases, not something to fabricate a value for
```

```python
# investigate the 3 orders with no payment record at all
print(master[master['total_payment_value'].isnull()][['order_id', 'order_status']])
```

```python
# ===== FEATURE 1: Delivery delay (only meaningful for delivered orders) =====
# negative = delivered early, positive = delivered late
master['delivery_delay_days'] = (master['order_delivered_customer_date'] - master['order_estimated_delivery_date']).dt.days

# ===== FEATURE 2: Order approval time in hours =====
master['approval_time_hours'] = (master['order_approved_at'] - master['order_purchase_timestamp']).dt.total_seconds() / 3600

# ===== FEATURE 3: Time-based columns for trend analysis =====
master['order_month'] = master['order_purchase_timestamp'].dt.to_period('M')
master['order_year'] = master['order_purchase_timestamp'].dt.year
master['order_weekday'] = master['order_purchase_timestamp'].dt.day_name()

# quick check - confirm these new columns look reasonable
master[['delivery_delay_days', 'approval_time_hours', 'order_month', 'order_weekday']].describe(include='all')
```

```python
# investigate the most extreme late-delivery outliers - are these real logistics failures or data errors?
master.sort_values('delivery_delay_days', ascending=False)[['order_id', 'order_status', 'order_purchase_timestamp', 
                                                                'order_delivered_customer_date', 'order_estimated_delivery_date', 
                                                                'delivery_delay_days']].head(10)
```

```python
# investigate extreme approval time outliers - is this a real business delay or bad data?
master.sort_values('approval_time_hours', ascending=False)[['order_id', 'order_status', 'order_purchase_timestamp', 
                                                                'order_approved_at', 'approval_time_hours']].head(10)
```

```python
# flag extreme approval delays rather than deleting them - these are real recorded events, not corrupted data
# threshold: anything beyond 3 days (72 hours) to approve is unusual and worth separating from "normal" orders
master['approval_delay_flag'] = master['approval_time_hours'] > 72

# check how common this actually is
print(master['approval_delay_flag'].value_counts())
print("Percent flagged:", round(master['approval_delay_flag'].mean() * 100, 2), "%")
```

```python
# same flagging approach for extreme delivery delays
master['delivery_delay_flag'] = master['delivery_delay_days'] > 30  # more than a month late
print(master['delivery_delay_flag'].value_counts())
```

# ===== RFM SEGMENTATION =====

```python
# ===== RFM SEGMENTATION =====
# RFM = Recency (how recently they bought), Frequency (how often), Monetary (how much they spent)
# must group by customer_unique_id, NOT customer_id (customer_id is unique per order, not per person)

# set a reference date - one day after the last purchase in the dataset (standard RFM practice)
reference_date = master['order_purchase_timestamp'].max() + pd.Timedelta(days=1)
print("Reference date for RFM:", reference_date)

# build RFM table - only use delivered/valid orders with real purchase dates
rfm = master.groupby('customer_unique_id').agg(
    recency_days=('order_purchase_timestamp', lambda x: (reference_date - x.max()).days),
    frequency=('order_id', 'nunique'),  # nunique because master is item-level, not order-level
    monetary=('total_payment_value', 'sum')
).reset_index()

# quick sanity check
print(rfm.shape)
rfm.describe()
```

```python
# check if total_payment_value duplicates itself across multiple item rows for the same order
sample_order = master[master['order_id'] == master['order_id'].value_counts().index[0]]
print(sample_order[['order_id', 'order_item_id', 'total_payment_value']])
```

```python
# STEP A: collapse master down to one row per order (order-level), 
# picking the first row per order_id since payment/customer/date info is identical across all item rows of that order
order_level = master.drop_duplicates(subset='order_id')[
    ['order_id', 'customer_unique_id', 'order_purchase_timestamp', 'total_payment_value']
]

# sanity check - order_level should now have far fewer rows than master, close to unique order count
print("master rows:", master.shape[0])
print("order_level rows:", order_level.shape[0])
print("unique order_ids in master:", master['order_id'].nunique())
```

```python
# rebuild RFM using order_level data - this fixes the monetary double-counting bug
reference_date = order_level['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

rfm = order_level.groupby('customer_unique_id').agg(
    recency_days=('order_purchase_timestamp', lambda x: (reference_date - x.max()).days),
    frequency=('order_id', 'nunique'),
    monetary=('total_payment_value', 'sum')
).reset_index()

print(rfm.shape)
rfm.describe()
```

```python
# ===== RFM SCORING =====
# score each dimension 1-5 (5 = best). Recency: lower is better, so we reverse the score direction.
# Monetary: standard quantile split works fine since it's continuous and varied.

# Recency score (lower recency = more recent = better = higher score)
rfm['R_score'] = pd.qcut(rfm['recency_days'], q=5, labels=[5,4,3,2,1])

# Monetary score (higher spend = better = higher score)
rfm['M_score'] = pd.qcut(rfm['monetary'], q=5, labels=[1,2,3,4,5])

# Frequency score - CANNOT use qcut here because most values are 1 (not enough spread for 5 quantiles)
# manual business-logic bins instead: 1 order = lowest score, 5+ orders = highest score
rfm['F_score'] = pd.cut(rfm['frequency'], bins=[0,1,2,3,5,rfm['frequency'].max()], 
                          labels=[1,2,3,4,5], include_lowest=True)

# combine into one RFM segment code, e.g. "555" = best possible customer
rfm['RFM_score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)

# quick check
rfm.head(10)
```

```python
# check how frequency actually distributes - confirm the one-time-buyer dominance
print(rfm['frequency'].value_counts().sort_index())
print("\nPercent one-time buyers:", round((rfm['frequency']==1).mean()*100, 2), "%")
```

```python
# ===== BUILD NAMED CUSTOMER SEGMENTS =====
# logic: repeat buyers (frequency >= 2) are rare and valuable on their own - treat them as a distinct group first
# then split one-time buyers by recency + monetary value

def assign_segment(row):
    # repeat customers are automatically high-value regardless of R/M, since repeat purchase itself is rare here
    if row['frequency'] >= 2:
        return 'Repeat Buyer'
    
    # for one-time buyers, use recency + monetary to differentiate value
    if row['R_score'] >= 4 and row['M_score'] >= 4:
        return 'Recent High-Value (One-time)'
    elif row['R_score'] >= 4 and row['M_score'] < 4:
        return 'Recent Low-Value (One-time)'
    elif row['R_score'] <= 2 and row['M_score'] >= 4:
        return 'Lapsed High-Value (One-time)'
    elif row['R_score'] <= 2 and row['M_score'] < 4:
        return 'Lost / Low Engagement'
    else:
        return 'Mid-Value (One-time)'

rfm['segment'] = rfm.apply(assign_segment, axis=1)

# check segment sizes and their business value
segment_summary = rfm.groupby('segment').agg(
    customer_count=('customer_unique_id', 'count'),
    avg_monetary=('monetary', 'mean'),
    total_revenue=('monetary', 'sum')
).sort_values('total_revenue', ascending=False)

print(segment_summary)
```

```python
# ===== SALES PERFORMANCE FEATURES =====

# monthly revenue trend - only count each order's payment once (avoid the same duplication bug as before)
monthly_revenue = order_level.merge(
    master[['order_id', 'order_month']].drop_duplicates(subset='order_id'),
    on='order_id', how='left'
).groupby('order_month')['total_payment_value'].sum().reset_index()

print(monthly_revenue)
```

```python
# category-level revenue - THIS one is fine at item-level, because price is genuinely per-item, not per-order
# (unlike total_payment_value which is order-level and gets duplicated across items)
category_revenue = master.groupby('product_category_name_english')['price'].sum().sort_values(ascending=False)

print(category_revenue.head(10))
```

```python
# seller performance - revenue generated + average delivery delay per seller
seller_performance = master.groupby('seller_id').agg(
    total_revenue=('price', 'sum'),
    order_count=('order_id', 'nunique'),
    avg_delivery_delay=('delivery_delay_days', 'mean'),
    avg_review_score=('review_score', 'mean')
).sort_values('total_revenue', ascending=False)

print(seller_performance.head(10))
```

```python
# ===== MERGE SEGMENTS BACK INTO ORDER-LEVEL DATA =====
# attach customer segment labels onto order_level so every order knows its customer's segment
order_level = order_level.merge(
    rfm[['customer_unique_id', 'recency_days', 'frequency', 'monetary', 'R_score', 'F_score', 'M_score', 'segment']],
    on='customer_unique_id', how='left'
)

# quick check - every order should now have a segment label
print(order_level['segment'].value_counts())
print("\nAny missing segments?", order_level['segment'].isnull().sum())
```

```python
# also attach segment info to the full item-level master (useful for category x segment analysis in Power BI later)
master = master.merge(
    rfm[['customer_unique_id', 'segment']],
    on='customer_unique_id', how='left'
)

print(master['segment'].isnull().sum())  # should be 0 or very close to it
```

```python
# ===== EXPORT CLEANED DATA =====
# export order-level (for SQL revenue/segment queries) and item-level (for category/product analysis)

master.to_csv('master_item_level.csv', index=False)
order_level.to_csv('master_order_level.csv', index=False)
rfm.to_csv('rfm_customer_segments.csv', index=False)

print("Files exported: master_item_level.csv, master_order_level.csv, rfm_customer_segments.csv")
```
